Use numpy to randomly generate the numerical columns with realistic constraints. For example, larger house have a higher price. Set a random seed for reproducibility.
```python
#Part a

import numpy as np
import pandas as pd

#Set random seed
np.random.seed(42)

#Number of rows
n = 30

#Generate size (base variable)
size = np.random.randint(40,300,n) #house sizes in sqm

#Bedrooms depend on size
bedrooms = (size // 50) + np.random.randint(0,2,n)
bedrooms = np.clip(bedrooms, 1, 6)

#Bathrooms depend on bedrooms
bathrooms = np.clip(bedrooms - 1 + np.random.randint(0,2,n), 1, 5)

#Year built
year_built = np.random.randint(2015,2024,n)

#Categorical data
locations = ["Ntinda", "Kira", "Naalya", "Mukono", "Entebbe"]
property_types =["apartment", "bungalow", "mansion"]

location = np.random.choice(locations,n)
property_type = np.random.choice(property_types, n)

#Price depends on size & property type & location
base_price = size * 1_200_00 #price per sqm

#adjust for type
type_multiplier = {
    "apartment": 0.9,
    "bungalow": 1.1,
    "mansion": 1.6
}

#adjust for location
location_multiplier = {
    "Ntinda": 1.3,
    "Kira": 1.1,
    "Naalya": 1.2,
    "Mukono": 0.8,
    "Entebbe": 1.5
}

price = []
for s, t, l in zip(size, property_type, location):
    p = s * 1_200_000
    p *= type_multiplier[t]
    p*= location_multiplier[l]
    price.append(int(p))

#CREATE DATAFRAME
df = pd.DataFrame({
    "price_ugx":price,
    "size_sqm": size,
    "bedrooms": bedrooms,
    "bathrooms": bathrooms,
    "location": location,
    "year_built":year_built,
    "property_type": property_type

})

#Display Result
print("Shape:", df.shape)
print("\nDtypes:\n", df.dtypes)
print("\nFirst 8 rows:\n", df.head(8))
```

```python
#Part b

filtered = df[
    ((df["location"] == "Ntinda") | (df["location"] == "Entebbe"))& 
    (df["price_ugx"] > 400000000) & 
    (df["bedrooms"] >= 3)
].reset_index(drop=True)

print(filtered)
```

```python
#Part c

#Slicing columns 2 and 3 and rows 5 to 14 inclusive
slice_df = df.iloc[5:15, [2,3]]

#renaming columns
slice_df.columns = ["house_size", "num_bedrooms"]

#Mean and std deviation
print("Mean:\n", slice_df.mean())
print("\nStandard deviation:\n", slice_df.std())

#rows with highest value in first column
max_row = slice_df.loc[slice_df["num_bedrooms"].idxmax()]
print("\nRow with highest bedrooms:\n", max_row)
```
